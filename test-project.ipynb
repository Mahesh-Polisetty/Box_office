```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# Load your dataset (adjust the path and file format as needed)
# dataset = pd.read_csv('path_to_your_data.csv')
# Assuming 'dataset' has 'company_names' and 'genre_names' split already, or:
dataset['genre_list'] = dataset['genre_names'].str.split(', ')
exploded_genres = dataset.explode('genre_list')

# Count movies by company and genre
company_genre_counts = exploded_genres.groupby(['company_names', 'genre_list']).size().reset_index(name='num_movies')

# Get top 10 genres and top 10 companies
top_genres = company_genre_counts.groupby('genre_list')['num_movies'].sum().nlargest(10).index
top_companies = company_genre_counts.groupby('company_names')['num_movies'].sum().nlargest(10).index

# Filter data for only top genres and companies
filtered_data = company_genre_counts[(company_genre_counts['genre_list'].isin(top_genres)) & (company_genre_counts['company_names'].isin(top_companies))]

# Create the graph
G = nx.Graph()

# Add nodes with the node attribute "bipartite"
genres = list(top_genres)
companies = list(top_companies)
G.add_nodes_from(genres, bipartite=0, node_color='lightblue', node_size=1000)
G.add_nodes_from(companies, bipartite=1, node_color='lightgreen', node_size=1000)

# Add edges with weights
for _, row in filtered_data.iterrows():
    G.add_edge(row['company_names'], row['genre_list'], weight=row['num_movies'])

# Draw the network graph
pos = nx.bipartite_layout(G, genres)
plt.figure(figsize=(12, 12))
nx.draw(G, pos, with_labels=True, node_color=[data['node_color'] for v, data in G.nodes(data=True)],
        node_size=[data['node_size'] for v, data in G.nodes(data=True)], font_size=8, edge_color='#AAAAAA')
labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=labels)
plt.title('Network Graph of Top 10 Genres and Top 10 Companies')
plt.show()
```

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from collections import defaultdict
import matplotlib

# Set an interactive backend
matplotlib.use('TkAgg')

# Assuming 'filtered_data' is already defined and loaded
# Create the graph
G = nx.Graph()

# Define nodes with the node attribute "bipartite"
companies = set(filtered_data['company_names'])
genres = set(filtered_data['genre_list'])
# Node size proportional to the number of movies
company_sizes = [filtered_data[filtered_data['company_names'] == company]['num_movies'].sum() * 10 for company in companies]
genre_sizes = [filtered_data[filtered_data['genre_list'] == genre]['num_movies'].sum() * 10 for genre in genres]

G.add_nodes_from(companies, bipartite=0, size=company_sizes)
G.add_nodes_from(genres, bipartite=1, size=genre_sizes)

# Add edges with weights proportional to the number of movies
for _, row in filtered_data.iterrows():
    G.add_edge(row['company_names'], row['genre_list'], weight=row['num_movies'])

# Layout to spread nodes and minimize overlap
pos = nx.spring_layout(G, k=0.15, iterations=20)

# Draw the network graph
plt.figure(figsize=(14, 14))
nx.draw_networkx_nodes(G, pos, nodelist=companies, node_color='navy', node_size=company_sizes, label='Companies')
nx.draw_networkx_nodes(G, pos, nodelist=genres, node_color='red', node_size=genre_sizes, label='Genres')
edge_widths = [G[u][v]['weight'] / 10 for u, v in G.edges()]
nx.draw_networkx_edges(G, pos, width=edge_widths, alpha=1, edge_color='black')
nx.draw_networkx_labels(G, pos, labels={node: node for node in companies.union(genres)}, font_size=8, font_color='black')

plt.title('Enhanced Network Graph of Top 20 Companies and Movie Genres', fontsize=16)
plt.legend(scatterpoints=1)
plt.axis('off')
plt.show()
```
